```
%py
spark.catalog.setCurrentCatalog("purgo_databricks")

# PySpark script: Generate comprehensive test data for purgo_playground.s3_file_process_log table
# Purpose: To create diverse test records covering happy path, edge, error, NULL, and special character scenarios for S3 file migration logic
# Author: Giang Nguyen
# Date: 2025-09-29
# Description: This script generates 30 test records for the s3_file_process_log table, including valid, boundary, error, NULL, and special/multibyte character cases. It ensures all data types and formats match Databricks standards, and covers all migration scenarios described in the requirements and design.

# from pyspark.sql import SparkSession  # SparkSession is already available in Databricks
from pyspark.sql.types import StructType, StructField, StringType, TimestampType  
from pyspark.sql import Row  
from pyspark.sql.functions import lit  

def get_test_data_rows():
    """
    Returns a list of Row objects representing diverse test data for s3_file_process_log.
    Covers happy path, edge, error, NULL, and special character scenarios.

    Returns:
        List[Row]: List of Row objects for DataFrame creation
    """
    return [
        # Happy path: Valid migration scenarios
        Row(file_name="patient_001.csv", s3_vendor_path="s3://vendor/purgo/patient_raw", s3_landing_path="s3://bucket/filestore/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="SUCCESS", file_processed_date="2024-06-01T10:00:00.000+0000"),
        Row(file_name="lab_202406.csv", s3_vendor_path="s3://vendor/purgo/lab_raw", s3_landing_path="s3://bucket/filestore/purgo/lab_raw", s3_archive_path="s3://bucket/filestore/purgo/lab_archive", file_status="SUCCESS", file_processed_date="2024-06-02T11:00:00.000+0000"),
        Row(file_name="visit_123.json", s3_vendor_path="s3://vendor/purgo/visit_raw", s3_landing_path="s3://bucket/filestore/purgo/visit_raw", s3_archive_path="s3://bucket/filestore/purgo/visit_archive", file_status="SUCCESS", file_processed_date="2024-06-03T12:00:00.000+0000"),
        Row(file_name="patient_002.csv", s3_vendor_path="s3://vendor/purgo/patient_raw", s3_landing_path="s3://bucket/filestore/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="SUCCESS", file_processed_date="2024-06-04T13:00:00.000+0000"),
        Row(file_name="imaging_abc.dcm", s3_vendor_path="s3://vendor/purgo/imaging_raw", s3_landing_path="s3://bucket/filestore/purgo/imaging_raw", s3_archive_path="s3://bucket/filestore/purgo/imaging_archive", file_status="SUCCESS", file_processed_date="2024-06-05T14:00:00.000+0000"),
        # Happy path: Duplicate file_name with different paths
        Row(file_name="patient_007.csv", s3_vendor_path="s3://vendor/purgo/patient_raw", s3_landing_path="s3://bucket/filestore/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="SUCCESS", file_processed_date="2024-06-15T08:00:00.000+0000"),
        Row(file_name="patient_007.csv", s3_vendor_path="s3://vendor/purgo/patient_raw2", s3_landing_path="s3://bucket/filestore/purgo/patient_raw2", s3_archive_path="s3://bucket/filestore/purgo/patient_archive2", file_status="SUCCESS", file_processed_date="2024-06-15T09:00:00.000+0000"),
        # Happy path: File already exists in archive folder
        Row(file_name="patient_008.csv", s3_vendor_path="s3://vendor/purgo/patient_raw", s3_landing_path="s3://bucket/filestore/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="SUCCESS", file_processed_date="2024-06-16T10:00:00.000+0000"),
        # Happy path: Special/multibyte characters in file_name
        Row(file_name="患者_009.csv", s3_vendor_path="s3://vendor/purgo/patient_raw", s3_landing_path="s3://bucket/filestore/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="SUCCESS", file_processed_date="2024-06-17T11:00:00.000+0000"),
        Row(file_name="lab_特殊字符.csv", s3_vendor_path="s3://vendor/purgo/lab_raw", s3_landing_path="s3://bucket/filestore/purgo/lab_raw", s3_archive_path="s3://bucket/filestore/purgo/lab_archive", file_status="SUCCESS", file_processed_date="2024-06-18T12:00:00.000+0000"),
        Row(file_name="visit_emoji_😀.json", s3_vendor_path="s3://vendor/purgo/visit_raw", s3_landing_path="s3://bucket/filestore/purgo/visit_raw", s3_archive_path="s3://bucket/filestore/purgo/visit_archive", file_status="SUCCESS", file_processed_date="2024-06-19T13:00:00.000+0000"),
        # Edge case: file_status not eligible
        Row(file_name="patient_003.csv", s3_vendor_path="s3://vendor/purgo/patient_raw", s3_landing_path="s3://bucket/filestore/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="FAILED", file_processed_date="2024-06-06T15:00:00.000+0000"),
        Row(file_name="lab_202407.csv", s3_vendor_path="s3://vendor/purgo/lab_raw", s3_landing_path="s3://bucket/filestore/purgo/lab_raw", s3_archive_path="s3://bucket/filestore/purgo/lab_archive", file_status="IN_PROGRESS", file_processed_date="2024-06-07T16:00:00.000+0000"),
        Row(file_name="visit_124.json", s3_vendor_path="s3://vendor/purgo/visit_raw", s3_landing_path="s3://bucket/filestore/purgo/visit_raw", s3_archive_path="s3://bucket/filestore/purgo/visit_archive", file_status="ERROR", file_processed_date="2024-06-08T17:00:00.000+0000"),
        # Error case: Missing required fields
        Row(file_name="patient_004.csv", s3_vendor_path="s3://vendor/purgo/patient_raw", s3_landing_path=None, s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="SUCCESS", file_processed_date="2024-06-09T18:00:00.000+0000"),
        Row(file_name="lab_202408.csv", s3_vendor_path="s3://vendor/purgo/lab_raw", s3_landing_path="s3://bucket/filestore/purgo/lab_raw", s3_archive_path=None, file_status="SUCCESS", file_processed_date="2024-06-10T19:00:00.000+0000"),
        Row(file_name="visit_125.json", s3_vendor_path="s3://vendor/purgo/visit_raw", s3_landing_path="s3://bucket/filestore/purgo/visit_raw", s3_archive_path="s3://bucket/filestore/purgo/visit_archive", file_status=None, file_processed_date="2024-06-11T20:00:00.000+0000"),
        # Error case: Invalid S3 path
        Row(file_name="patient_005.csv", s3_vendor_path="s3://vendor/purgo/patient_raw", s3_landing_path="s3://invalid-bucket/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="SUCCESS", file_processed_date="2024-06-12T21:00:00.000+0000"),
        Row(file_name="lab_202409.csv", s3_vendor_path="s3://vendor/purgo/lab_raw", s3_landing_path="s3://bucket/filestore/purgo/lab_raw", s3_archive_path="s3://invalid-bucket/purgo/lab_archive", file_status="SUCCESS", file_processed_date="2024-06-13T22:00:00.000+0000"),
        # Error case: Unsupported file type
        Row(file_name="patient_006.exe", s3_vendor_path="s3://vendor/purgo/patient_raw", s3_landing_path="s3://bucket/filestore/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="SUCCESS", file_processed_date="2024-06-14T23:00:00.000+0000"),
        # Edge case: file_processed_date far future
        Row(file_name="future_001.csv", s3_vendor_path="s3://vendor/purgo/patient_raw", s3_landing_path="s3://bucket/filestore/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="SUCCESS", file_processed_date="2099-12-31T23:59:59.999+0000"),
        # Edge case: file_processed_date far past
        Row(file_name="past_001.csv", s3_vendor_path="s3://vendor/purgo/patient_raw", s3_landing_path="s3://bucket/filestore/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="SUCCESS", file_processed_date="1900-01-01T00:00:00.000+0000"),
        # Edge case: file_name with special symbols
        Row(file_name="file_!@#$.csv", s3_vendor_path="s3://vendor/purgo/patient_raw", s3_landing_path="s3://bucket/filestore/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="SUCCESS", file_processed_date="2024-06-20T14:00:00.000+0000"),
        # Edge case: file_name with spaces
        Row(file_name="file with spaces.csv", s3_vendor_path="s3://vendor/purgo/patient_raw", s3_landing_path="s3://bucket/filestore/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="SUCCESS", file_processed_date="2024-06-21T15:00:00.000+0000"),
        # Edge case: file_name with very long name
        Row(file_name="a"*255 + ".csv", s3_vendor_path="s3://vendor/purgo/patient_raw", s3_landing_path="s3://bucket/filestore/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="SUCCESS", file_processed_date="2024-06-22T16:00:00.000+0000"),
        # Edge case: s3_vendor_path is NULL
        Row(file_name="vendor_null.csv", s3_vendor_path=None, s3_landing_path="s3://bucket/filestore/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="SUCCESS", file_processed_date="2024-06-23T17:00:00.000+0000"),
        # Edge case: s3_vendor_path with special characters
        Row(file_name="vendor_special.csv", s3_vendor_path="s3://vendor/purgo/特殊字符", s3_landing_path="s3://bucket/filestore/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="SUCCESS", file_processed_date="2024-06-24T18:00:00.000+0000"),
        # Edge case: s3 paths with trailing slash
        Row(file_name="trailing_slash.csv", s3_vendor_path="s3://vendor/purgo/patient_raw/", s3_landing_path="s3://bucket/filestore/purgo/patient_raw/", s3_archive_path="s3://bucket/filestore/purgo/patient_archive/", file_status="SUCCESS", file_processed_date="2024-06-25T19:00:00.000+0000"),
        # Edge case: s3 paths with uppercase letters
        Row(file_name="uppercase_path.csv", s3_vendor_path="s3://VENDOR/PURGO/PATIENT_RAW", s3_landing_path="s3://BUCKET/FILESTORE/PURGO/PATIENT_RAW", s3_archive_path="s3://BUCKET/FILESTORE/PURGO/PATIENT_ARCHIVE", file_status="SUCCESS", file_processed_date="2024-06-26T20:00:00.000+0000"),
        # Edge case: file_status is lower case 'success'
        Row(file_name="lowercase_status.csv", s3_vendor_path="s3://vendor/purgo/patient_raw", s3_landing_path="s3://bucket/filestore/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="success", file_processed_date="2024-06-27T21:00:00.000+0000"),
        # Edge case: file_status is mixed case 'Success'
        Row(file_name="mixedcase_status.csv", s3_vendor_path="s3://vendor/purgo/patient_raw", s3_landing_path="s3://bucket/filestore/purgo/patient_raw", s3_archive_path="s3://bucket/filestore/purgo/patient_archive", file_status="Success", file_processed_date="2024-06-28T22:00:00.000+0000"),
        # Edge case: All fields NULL
        Row(file_name=None, s3_vendor_path=None, s3_landing_path=None, s3_archive_path=None, file_status=None, file_processed_date=None),
    ]

def get_s3_file_process_log_schema():
    """
    Returns the schema for s3_file_process_log table using Databricks native data types.

    Returns:
        StructType: Schema definition for DataFrame
    """
    return StructType([
        StructField("file_name", StringType(), True),
        StructField("s3_vendor_path", StringType(), True),
        StructField("s3_landing_path", StringType(), True),
        StructField("s3_archive_path", StringType(), True),
        StructField("file_status", StringType(), True),
        StructField("file_processed_date", TimestampType(), True),
    ])

def convert_rows_to_df(rows, schema):
    """
    Converts a list of Row objects to a PySpark DataFrame with the given schema.
    Ensures timestamp fields are properly converted to TimestampType.

    Args:
        rows (List[Row]): List of Row objects
        schema (StructType): Schema for DataFrame

    Returns:
        DataFrame: PySpark DataFrame with validated schema
    """
    from pyspark.sql import DataFrame  
    import datetime  
    def parse_timestamp(ts):
        if ts is None:
            return None
        # Accepts 'YYYY-MM-DDTHH:MM:SS.sss+0000' or 'YYYY-MM-DDTHH:MM:SSZ'
        try:
            if ts.endswith('Z'):
                return datetime.datetime.strptime(ts, "%Y-%m-%dT%H:%M:%SZ")
            elif '+' in ts:
                # Remove timezone for Spark compatibility
                ts_main = ts.split('+')[0]
                return datetime.datetime.strptime(ts_main, "%Y-%m-%dT%H:%M:%S.%f")
            else:
                return datetime.datetime.strptime(ts, "%Y-%m-%dT%H:%M:%S.%f")
        except Exception:
            return None

    # Convert timestamps to datetime objects for Spark
    converted = []
    for r in rows:
        d = r.asDict()
        d['file_processed_date'] = parse_timestamp(d.get('file_processed_date'))
        converted.append(Row(**d))
    return spark.createDataFrame(converted, schema)

def write_test_data_to_catalog(df, catalog_table):
    """
    Writes the test DataFrame to the specified Unity Catalog table.
    Overwrites existing data for repeatable test runs.

    Args:
        df (DataFrame): PySpark DataFrame with test data
        catalog_table (str): Fully qualified Unity Catalog table name

    Returns:
        None
    """
    df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable(catalog_table)

def main():
    """
    Main function to generate and write test data for s3_file_process_log table.
    Handles file opening errors gracefully.
    """
    try:
        rows = get_test_data_rows()
        schema = get_s3_file_process_log_schema()
        df = convert_rows_to_df(rows, schema)
        write_test_data_to_catalog(df, "purgo_playground.s3_file_process_log")
        print("Test data successfully written to purgo_playground.s3_file_process_log")
    except Exception as e:
        print(f"Error generating or writing test data: {str(e)}")

if __name__ == "__main__":
    main()

# spark.stop()  # Do not stop SparkSession in Databricks
```
